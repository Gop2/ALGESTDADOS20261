```
#Faça um programa que peça o nome do usuário e exiba uma mensagem de boas-vindas com o nome digitado.

nome = input("Digite seu nome:")

print(f"Olá {nome} seja bem vindo")
```

```
#Desenvolva um programa que peça dois números ao usuário e mostre a soma entre eles.

numero1 = int(input("Digite um numero: "))
numero2 = int(input("Digite outro numero: "))

soma = numero1 + numero2

print(f"A soma dos numeros é {soma}")
```

```
#Crie um programa que peça um número inteiro e mostre seu antecessor e seu sucessor.

numero_inteiro = int(input("Digite um número inteiro: "))

antecessor = numero_inteiro - 1
sucessor = numero_inteiro + 1

print(f"O antecessor de {numero_inteiro} é {antecessor}.")
print(f"O sucessor de {numero_inteiro} é {sucessor}.")
```
