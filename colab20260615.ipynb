```
# A função map() aplica uma transformação a todos os elementos de um iterável sem laços explícitos. Solicite ao usuário 6 números inteiros e armazene-os em uma lista. Utilize map() com uma lambda para gerar uma nova lista com o dobro de cada valor. Exiba a lista original e a transformada lado a lado.

lista = []

for i in range(6):
    lista.append(int(input("Digite o valor:  ")))

new_list = list(map(lambda x: x*2, lista))

print(new_list)
```

```
# Em sistemas de monitoramento internacional, conversões de unidade são aplicadas a conjuntos de dados. Dado uma lista de temperaturas em Celsius fornecida pelo usuário, utilize map() com a fórmula F = C × 9/5 + 32 para converter tod
gelado_ou_frio = []

for i in range(5):
    gelado_ou_frio.append(int(input("Digite uma temperatura em Celsus: ")))

farihait = list(map(lambda x: round(x * (9/5) + 32), gelado_ou_frio))

print(f"\nTemperatura em celsus: {gelado_ou_frio}")
print(f"Temperatura em Fahrenheit: {farihait}")
```
