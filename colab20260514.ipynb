```
# Leia cinco nomes de produtos digitados pelo usuário e armazene-os em um conjunto. Ao final, exiba todos os produtos cadastrados e a quantidade total de itens únicos.

conjunto = set()

for i in range(5):
    item = input(f"digite o {i + 1} produto: ")
    conjunto.add(item)

print(f"\nProdutos cadastrados: {conjunto}")
print(f"Quantidade total de itens únicos: {len(conjunto)}")
```

```
# Crie uma tupla com nomes de cidades. Leia uma cidade digitada pelo usuário e verifique se ela existe na tupla. Exiba uma mensagem informando se a cidade foi encontrada ou não.

cidade = ("São Paulo", "Rio de Janeiro", "Belo Horizonte", "Porto Alegre", "Curitiba")

cidades = input("Digite uma cidade: ")

if cidades in cidade:
    print(f"A cidade {cidades} foi encontrada na tupla.")
else:
    print(f"A cidade {cidades} não foi encontrada na tupla.")
```
