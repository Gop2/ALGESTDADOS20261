```
#Usando for e range(), exiba os números de 1 até 10, um por linha.

for i in range(1,11):
    print(i)
```

```
#Usando for e range(), exiba os números de 10 até 1 em ordem decrescente, um por linha.

for i in range(10,0,-1):
    print(i)
```
