```
#Leia a nota de um aluno. Classifique usando elif: nota ≥ 9 → "A — Excelente", ≥ 7 → "B — Aprovado", ≥ 5 → "C — Recuperação", abaixo → "D — Reprovado".

nota = float(input("Digite a nota do aluno: "))

if nota >= 9:
    print("A - Excelente")
elif nota >= 7:
    print("B - Aprovado")
elif nota >= 5:
    print("C - Recuperação")
elif nota <= 4:
    print("D - Reprovado")
```

```
#Leia um número de 1 a 7 e exiba o nome do dia correspondente (1 = Segunda, ..., 7 = Domingo). Para qualquer outro número, exiba "Número inválido".

dia = int(input("Digite um numero de 1 a 7: "))

if dia == 1:
    print("segunda")
elif dia == 2:
    print("terça")
elif dia == 3:
    print("quarta")
elif dia == 4:
    print("quinta")
elif dia == 5:
    print("sexta")
elif dia == 6:
    print("sabado")
elif dia == 7:
    print("domingo")
else:
    print("nao existe este numero")
```

```
#Leia uma nota. Valide se está entre 0 e 10 usando encadeamento de comparações (0 <= nota <= 10). Se inválida, exiba "Nota inválida!". Se válida, exiba "Nota registrada: [nota]".

nota = float(input("digite sua nota: "))

if 0 <= nota <= 10:
    print(f"nota registrada {nota}")
    print("nota valida")
else:
    print("nota invalida")

```

```
#Leia a idade. Classifique: menor de 12 → "Criança", 12 a 17 → "Adolescente", 18 a 59 → "Adulto", 60 ou mais → "Idoso". Valide que a idade não é negativa.

idade = int(input("digite a idade: "))

if idade < 0 or idade > 120:
    print("idade invalida")
elif idade < 12:
    print("crianca")
elif idade <= 17:
    print("adolescente")
elif idade <= 59:
    print("adulto")
else:
    print("idoso")
```
