```
# Crie dois conjuntos representando alunos presentes em dois dias diferentes. Exiba quais alunos compareceram apenas no primeiro dia utilizando a operação de diferença.


primeiro = {"ze", "tio san", "ben 10", "ranger vermelho"}

segundo = {"ze", "avatar", "ranger vermelho", "5 hokage"}

print(f"compareceram no primeiro dia: {primeiro.difference(segundo)}")
```

```
# Crie uma tupla com 8 números inteiros. Leia um número e verifique em quais posições ele aparece dentro da tupla. Caso não exista, exiba uma mensagem apropriada.

numero = (2,23,44,5,66,7,7,67)

verificar = int(input("digite um numero: "))

found = False
for posi in range(len(numero)):
    if verificar == numero[posi]:
        print(f"O número {verificar} está na posição {posi}")
        found = True

if not found:
    print(f"O número {verificar} não está na tupla")
```

```
# Crie uma tupla com os nomes de 5 times de futebol. Exiba o primeiro colocado, o último colocado e a lista em ordem alfabética.

maiores_times_do_brasil = ("Vitoria", "Al- mossar", "Bangu", "Serra Fc", "Barceloma de Quayaquil")

print(f"Primeiro colocado: {maiores_times_do_brasil[0]}")
print(f"Último colocado: {maiores_times_do_brasil[-1]}")
print(f"Times em ordem alfabética: {sorted(maiores_times_do_brasil)}")
```

```
#Crie uma tupla com 10 números inteiros. Exiba a soma total, a média dos valores, o maior número, o menor número e quantos valores são pares.


numeros = (1,2,3,4,5,6,7,8,10)

pares = 0

for num in numeros:
    if num % 2 == 0:
        pares += 1

maior = max(numeros)
menor = min(numeros)
soma = sum(numeros)
media = soma / len(numeros)

print(f'Soma Total: {soma}\nMédia: {media}\nMaior valor: {maior}\nMenor valor:{menor}\nPares: {pares}')
```
