```
# Crie uma lista com 4 nomes de fases. Use enumerate() dentro de um for para exibir o índice e o nome de cada fase (Ex: "Fase 0: Castelo de Fogo").

Rivaldo = ["Max Steel Verde", "Irmão do Jorel", "All Star Tower Defense", "Neymar vs Robinho Jr"]


for index, name in enumerate(Rivaldo):
    print(f"Fase {index}: {name}")
```
