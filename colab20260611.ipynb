```
def saudacao(nome):
    mensagem = "Olá, " + nome + "!"
    return mensagem

resultado = saudacao("Maria")
print(resultado)
```

```
def neymar(nome):
    mensagem = "Bom, o top 3 global de todos os tempos é o " + nome + "!"
    return mensagem

resultado = neymar("Garrincha")
print(resultado)
```

```
def soma(a, b):
    resultado = (a + b)
    return resultado

print(soma(5, 8))
print(soma(67, 67))
print(soma(100000000000000000000000, 245376878156527916))
print(soma(-1, 22))
```

```
def quadrado(n):
    return n ** 2

x = quadrado(5)
print(x)
print(quadrado(3))
```
