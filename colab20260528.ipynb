```
#Dado o dicionário produto = {"nome": "Caderno", "preco": 15.90, "estoque": 42}, escreva um programa que exiba separadamente o nome do produto, o preço e a quantidade em estoque. Use o acesso direto por colchetes para ao menos um valor e o método .get() para ao menos um outro.

keep = {
    "nome":  'Caderno',
    "preco": 15.90,
    "estoque": 42
}

nome_produto = keep["nome"]
preco_produto = keep["preco"]
estoque_produto = keep["estoque"]

print(f"Nome do produto: {nome_produto}")
print(f"Preço do produto: {preco_produto}")
print(f"Quantidade em estoque: {estoque_produto}")
```

```
#Dado um dicionário de agenda telefônica com pelo menos 5 contatos (onde cada chave é um nome e o valor é um número de telefone), exiba: (a) quantos contatos há na agenda usando len(); (b) todos os nomes usando .keys(); (c) todos os telefones usando .values().


agenda = {
    "Dembele": "999292919",
    "Gustavo Alves meu macaco favorito": "676767667676767676767676767",
    "negoney": "666",
    "messi arabe": "8000 2222",
    "general divino mahoraga": "01010011 01001111 01010011"

}

print(f"Quantidade de contatos na agenda: {len(agenda)}")
print()

print("Nomes na agenda:")
for nome in agenda.keys():
    print(nome)
    print()

print("Telefones na agenda:")
for telefone in agenda.values():
    print(telefone)
    print()

for chave, valor in agenda.items():
    print(f"{chave}: {valor}")



```
